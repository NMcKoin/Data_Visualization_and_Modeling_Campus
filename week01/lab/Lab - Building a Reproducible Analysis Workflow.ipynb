## Step 3: Operation

Create one Boolean condition that identifies Canaveral National Seashore and another that identifies July.


```python
# Write your code here.
is_canaveral = park_visits["park"] == "Canaveral National Seashore"
is_july = park_visits["month"] == "July"
```

Combine the two conditions with `&`. Use `.loc` to select the `year` and `recreation_visits` features, make a copy named `canaveral_july`, and display it.


```python
# Write your code here.
combinedCond = is_canaveral & is_july
canaveral_july = park_visits.loc[combinedCond, ["year", "recreation_visits"]].copy()
canaveral_july
```

Create `pre_pandemic_condition`, which is `True` for years from 2015 through 2019. Use it with `.loc` to calculate the mean of `recreation_visits`, and store the result as `pre_pandemic_july_mean`.

Display `pre_pandemic_july_mean`.


```python
# Write your code here.
pre_pandemic_condition = (park_visits["year"] >= 2015) & (park_visits["year"] <= 2019)
pre_pandemic_july_mean = park_visits.loc[combinedCond & pre_pandemic_condition, "recreation_visits"].mean()
print(pre_pandemic_july_mean)
```

Create a condition that identifies rows from 2020 onward whose `recreation_visits` are greater than or equal to `pre_pandemic_july_mean`. Use `.loc` to create `recovery_years`, then display it in ascending year order.


```python
# Write your code here.
recovCond = is_canaveral & is_july & (park_visits["year"] >= 2020) & (park_visits["recreation_visits"] >= pre_pandemic_july_mean)
recovery_years = park_visits.loc[recovCond].sort_values("year")
print(recovery_years)
```

## Step 3: Operation

Write the code needed to prepare, filter, group, summarize, or sort the data. Store the final evidence in a clearly named object and display it.


```python
# Write your code here.
#check firstparty
n64Games['FirstParty'] = n64Games['Developer(s)'].str.contains('Nintendo EAD|HAL Laboratory', case=False, regex=True)
#Group and summarize
sales_summary = n64Games.groupby('FirstParty')['Sales'].agg(['count', 'mean', 'median', 'sum'])
#Renaming for readability
sales_summary = sales_summary.rename(columns={'count': 'Titles'})
sales_summary = sales_summary.rename(columns={'sum': 'Units Sold'})
#Sort
sales_summary = sales_summary.sort_values('mean', ascending=False)

#formating for readability
pd.set_option('display.float_format', '{:.0f}'.format)
final = sales_summary
final
```
